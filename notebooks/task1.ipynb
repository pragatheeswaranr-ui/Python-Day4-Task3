## Introduction

This notebook is created to practise Python basics and compare execution time.

```python
numbers = list(range(1000))

%timeit [x * 2 for x in numbers]
```

```python
numbers = list(range(1000))

def double_numbers():
    result = []
    for x in numbers:
        result.append(x * 2)
    return result

%timeit double_numbers()
```
